```python
import numpy as np
import pandas as pd 
```

```python
df=pd.read_csv("messy_ecommerce_sales_dataset.csv")
df.head()
```

```python
df.describe
```

```python
df.info
```

```python
df.isnull().sum()
```

```python
# Text columns ka whitespace trim karo
df['CustomerName'] = df['CustomerName'].str.strip()
df['Email'] = df['Email'].str.strip()
df['ProductName'] = df['ProductName'].str.strip()
df['Category'] = df['Category'].str.strip()
df['PaymentMethod'] = df['PaymentMethod'].str.strip()
df['OrderStatus'] = df['OrderStatus'].str.strip()
df['ShippingCity'] = df['ShippingCity'].str.strip()
df['Country'] = df['Country'].str.strip()

df.replace(['N/A', 'NA', 'None', ''], np.nan, inplace=True)

```

```python

df['Category'].unique()
```

```python
df['Category'] = df['Category'].str.title()
df['Category'].unique()
```

```python
df['PaymentMethod'] = df['PaymentMethod'].str.upper()
df['PaymentMethod'] = df['PaymentMethod'].replace({
    'COD': 'Cash on Delivery',
    'CASH ON DELIVERY': 'Cash on Delivery',
    'WALLET': 'Wallet',
    'NET BANKING': 'Net Banking',
    'DEBIT CARD': 'Debit Card',
    'CREDIT CARD': 'Credit Card',
    'UPI': 'UPI',
    'PAYPAL': 'PayPal',
    'EMI': 'EMI'
})
df
```

```python
df['OrderStatus'] = df['OrderStatus'].str.upper()
df['OrderStatus'] = df['OrderStatus'].replace({
    'PENDING': 'Pending',
    'PROCESSING': 'Processing',
    'SHIPPED': 'Shipped',
    'DELIVERED': 'Delivered',
    'DELVERED': 'Delivered',
    'CANCELLED': 'Cancelled',
    'RETURNED': 'Returned',
    'IN TRANSIT': 'In Transit'
})
df
```

```python
print(df['PaymentMethod'].unique())
print(df['OrderStatus'].unique())
```

```python
df.duplicated(subset=['OrderID']).sum()
```

```python
df = df.drop_duplicates(subset=['OrderID'], keep='first')
df
```

```python
print(df.shape)
print(df.duplicated(subset=['OrderID']).sum())
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'], format='mixed', dayfirst=True, errors='coerce')
df['ShipDate'] = pd.to_datetime(df['ShipDate'], format='mixed', dayfirst=True, errors='coerce')
```

```python
print(df['OrderDate'].dtype)
print(df['OrderDate'].isnull().sum())
print(df[['OrderDate', 'ShipDate']].head(10))
```

```python
df.loc[df['ShipDate'] < df['OrderDate'], 'ShipDate'] = pd.NaT
```

```python
df['OrderDate'].isnull().sum()
```

```python
df_original = pd.read_csv('messy_ecommerce_sales_dataset.csv')
mask = df['OrderDate'].isnull() & df_original['OrderDate'].notnull()
df_original.loc[mask, 'OrderDate'].unique()[:20]
```

```python
df['Quantity'] = df['Quantity'].abs()
df
```

```python
df.loc[df['DiscountPercent'] > 100, 'DiscountPercent'] = np.nan
df
```

```python
print(df['Quantity'].min())
print(df['DiscountPercent'].max())
print(df['DiscountPercent'].isnull().sum())
```

```python
df.loc[df['Quantity'] == 0, 'Quantity'] = np.nan
print(df['Quantity'].min())
print(df['Quantity'].isnull().sum())
```

```python
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['DiscountPercent'] = df['DiscountPercent'].fillna(df['DiscountPercent'].median())
df['UnitPrice'] = df.groupby('ProductName')['UnitPrice'].transform(lambda x: x.fillna(x.median()))
df['UnitPrice'] = df['UnitPrice'].fillna(df['UnitPrice'].median())
df
```

```python
print(df['Quantity'].isnull().sum())
print(df['DiscountPercent'].isnull().sum())
print(df['UnitPrice'].isnull().sum())
```

TotalAmount = Quantity × UnitPrice × (1 - Discount/100)

```python
df['TotalAmount'] = df['Quantity'] * df['UnitPrice'] * (1 - df['DiscountPercent'] / 100)
df['TotalAmount'] = df['TotalAmount'].round(2)
```

```python
print(df['TotalAmount'].min())
print(df['TotalAmount'].isnull().sum())
df[['Quantity', 'UnitPrice', 'DiscountPercent', 'TotalAmount']].head(10)
```

```python
df[df['Email'].notnull() & ~df['Email'].str.contains('@', na=False)]['Email'].head(10)
```

```python
domains = ['gmail.com', 'yahoo.com', 'hotmail.com', 'outlook.com', 'rediffmail.com']

for d in domains:
    mask = df['Email'].notnull() & ~df['Email'].str.contains('@', na=False) & df['Email'].str.endswith(d)
    df.loc[mask, 'Email'] = df.loc[mask, 'Email'].str.replace(d, '@' + d, regex=False)
```

```python
broken_emails_after = df[df['Email'].notnull() & ~df['Email'].str.contains('@', na=False)]
print(broken_emails_after.shape)
df[df['Email'].str.contains('@', na=False)]['Email'].head(15)
```

```python
df['Category'] = df['Category'].fillna('Unknown')
df['PaymentMethod'] = df['PaymentMethod'].fillna('Unknown')
df['OrderStatus'] = df['OrderStatus'].fillna('Unknown')
df['ShippingCity'] = df['ShippingCity'].fillna('Unknown')
df['Country'] = df['Country'].fillna('Unknown')
df['CustomerName'] = df['CustomerName'].fillna('Unknown')
df['ProductName'] = df['ProductName'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df.to_csv('cleaned_sales_data.csv', index=False)
print("Saved successfully! Shape:", df.shape)
```
